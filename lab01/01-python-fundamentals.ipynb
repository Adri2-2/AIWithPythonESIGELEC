## 3. Lists, indexing, slicing, and dictionaries (7 min)

Python **lists** are mutable sequences. Indexing starts at zero, and a negative index counts from the end. A slice `values[start:stop]` includes `start` but *excludes* `stop`.

Python **dictionaries** associate keys with values, somewhat like Java's `Map`.

```python
temperatures = [18.2, 20.0, 21.5, 19.8, 23.1]
print("First:", temperatures[0])
print("Last:", temperatures[-1])
print("First three:", temperatures[:3])
print("Observations 2–4:", temperatures[1:4])

sample = {"sensor": "A1", "temperature": 22.5, "valid": True}
print("Sensor:", sample["sensor"])
sample["temperature"] = 23.0
print(sample)
```

## 4. Conditions, loops, and comprehensions (7 min)

Conditions and loops end with a colon (`:`); indentation replaces Java's braces. `enumerate()` gives both an index and a value. A **list comprehension** offers a concise way to transform or filter a sequence.

```python
for index, temp in enumerate(temperatures):
    if temp >= 21.0:
        print(f"Reading {index}: warm ({temp} °C)")
    else:
        print(f"Reading {index}: cool ({temp} °C)")

# List comprehension
fahrenheit = [temp * 9 / 5 + 32 for temp in temperatures]
warm_temperatures = [temp for temp in temperatures if temp >= 21.0]
print("Fahrenheit:", fahrenheit)
print("Warm readings:", warm_temperatures)
```

## 5. Functions and imports (5 min)

Define functions using `def`. Python does not require type declarations, but *type hints* are useful for readability. They are not automatically enforced at runtime.

We can import functionality from the Python standard library without installing extra packages.
The import can be done at the beginning of the Python file, or at any place where the library is needed.

```python
def celsius_to_fahrenheit(celsius: float) -> float:
    """Convert degrees Celsius into degrees Fahrenheit."""
    return celsius * 9 / 5 + 32

converted = [celsius_to_fahrenheit(t) for t in temperatures]
print("Converted:", converted)

import statistics
print("Average °C:", statistics.mean(temperatures))
```

### Exercise 4 — Reusable analysis (3 min)
Write a function `temperature_range(values)` returning the difference between the maximum and minimum observations. Test it using `temperatures`.

**Check:** The expected range is approximately **4.9 °C**.

*Hint:* `max(values) - min(values)`. Assume that the input list is not empty.

## 6. Mini-challenge — Explore a tiny dataset (5–8 min)

Real datasets consist of many observations, each with several attributes. Before we introduce Pandas, let's model this using a **list of dictionaries**. Each dictionary represents a row.

**Your task:**
1. Extract the temperatures of all *valid* readings.
2. Calculate their average using `statistics.mean()`.
3. Find the sensor name associated with the highest valid temperature.
4. Print a short conclusion: how many valid readings were available, what their mean temperature was, and which sensor had the highest temperature.

**Check:** There are **4 valid readings**; their mean is **21.45 °C**; the highest valid temperature belongs to sensor **B2**.

**Stretch task (optional):** What happens to the mean if you include the invalid reading? Why does data quality matter for machine learning?

```python
import statistics

readings = [
    {"sensor": "A1", "temperature": 19.5, "valid": True},
    {"sensor": "A2", "temperature": 21.0, "valid": True},
    {"sensor": "B1", "temperature": 99.9, "valid": False},
    {"sensor": "B2", "temperature": 24.1, "valid": True},
    {"sensor": "B3", "temperature": 21.2, "valid": True},
]

# Filter valid readings into a new list.
valid_readings = [r for r in readings if r["valid"]]
valid_temperatures = [r["temperature"] for r in valid_readings]

# Calculate mean valid temperature.
mean_temperature = statistics.mean(valid_temperatures)

# Identify the sensor with the maximum valid temperature.
hottest = max(valid_readings, key=lambda r: r["temperature"])

# Print your conclusion.
print(
    f"{len(valid_readings)} valid readings were available; "
    f"their mean temperature was {mean_temperature:.2f} °C; "
    f"the highest temperature ({hottest['temperature']} °C) came from sensor {hottest['sensor']}."
)

# Stretch task: include the invalid reading.
all_mean = statistics.mean(r["temperature"] for r in readings)
print(f"Mean including the invalid reading: {all_mean:.2f} °C")
```
