## 1. Python and notebook environment

We expect **Python 3.11 or newer** and, ideally, a Python interpreter inside the repository's `.venv` directory. A system interpreter may work, but usually means the wrong Jupyter kernel is selected.


````python
import sys
from pathlib import Path
import platform

print("Python version:", platform.python_version())
print("Python executable:", sys.executable)
print("Operating system:", platform.system(), platform.release())
print("Current directory:", Path.cwd())

python_ok = sys.version_info >= (3, 11)
venv_ok = Path(sys.prefix).resolve() != Path(sys.base_prefix).resolve()
print("Python >= 3.11:", "PASS" if python_ok else "CHECK")
print("Running in virtual environment:", "PASS" if venv_ok else "CHECK")
````

## 2. Required Python packages

The early labs use **NumPy**, **Pandas**, **Matplotlib**, and **scikit-learn**. JupyterLab and IPython are checked too. Each package is checked separately so a missing installation does not stop the notebook.


````python
import importlib
from importlib.metadata import version, PackageNotFoundError

packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "sklearn": "scikit-learn",
    "jupyterlab": "jupyterlab",
    "IPython": "ipython",
}
package_status = {}
for module_name, distribution_name in packages.items():
    try:
        importlib.import_module(module_name)
        installed_version = version(distribution_name)
        package_status[module_name] = True
        print(f"PASS  {distribution_name}: {installed_version}")
    except (ImportError, PackageNotFoundError, OSError) as error:
        package_status[module_name] = False
        print(f"FAIL  {distribution_name}: {type(error).__name__}: {error}")
````

## 3. NumPy — numerical calculation

Create an array and calculate its mean. The expected result is **3.0**.


````python
numpy_ok = False
try:
    import numpy as np
    numbers = np.array([1, 2, 3, 4, 5])
    result = float(numbers.mean())
    numpy_ok = np.isclose(result, 3.0)
    print("Array:", numbers)
    print("Mean:", result)
    print("NumPy:", "PASS" if numpy_ok else "FAIL")
except Exception as error:
    print("NumPy: FAIL -", error)
````

## 4. Pandas — tabular data and CSV

Create a small table, save a temporary CSV file, and read it back. This tests both Pandas and basic filesystem access. The temporary file is removed automatically.


````python
pandas_ok = False
csv_ok = False
try:
    import pandas as pd
    import tempfile
    sample = pd.DataFrame({"hours_studied": [1, 2, 3], "score": [60, 72, 85]})
    pandas_ok = sample.shape == (3, 2) and list(sample.columns) == ["hours_studied", "score"]
    with tempfile.TemporaryDirectory() as tmp:
        path = Path(tmp) / "environment-check.csv"
        sample.to_csv(path, index=False)
        restored = pd.read_csv(path)
        csv_ok = restored.equals(sample)
    display(restored)
    print("Pandas:", "PASS" if pandas_ok else "FAIL")
    print("CSV read/write:", "PASS" if csv_ok else "FAIL")
except Exception as error:
    print("Pandas / CSV: FAIL -", error)
````

## 5. Matplotlib — visualization

You should see a simple line chart below. If the chart is missing, ask for assistance even if the cell prints `PASS`.


````python
plot_ok = False
try:
    import matplotlib.pyplot as plt
    fig, ax = plt.subplots(figsize=(5, 3))
    ax.plot([1, 2, 3], [60, 72, 85], marker="o")
    ax.set(title="Environment check: sample scores", xlabel="Hours studied", ylabel="Score")
    ax.grid(alpha=0.3)
    plt.show()
    plot_ok = True
    print("Matplotlib: PASS (confirm that you can see the plot)")
except Exception as error:
    print("Matplotlib: FAIL -", error)
````

## 6. scikit-learn — train a first simple model

This is only a technical test; we will explain the model and its evaluation in a later lab.


````python
sklearn_ok = False
try:
    from sklearn.linear_model import LinearRegression
    X = [[1], [2], [3], [4]]
    y = [2, 4, 6, 8]
    model = LinearRegression().fit(X, y)
    prediction = float(model.predict([[5]])[0])
    sklearn_ok = abs(prediction - 10.0) < 1e-8
    print("Prediction for input 5:", round(prediction, 3))
    print("scikit-learn:", "PASS" if sklearn_ok else "FAIL")
except Exception as error:
    print("scikit-learn: FAIL -", error)
````

## 7. Final check

All package checks and practical tests should show `PASS`. A virtual environment is strongly recommended, but its absence is reported as a warning rather than failing the exercises.


````python
checks = {
    "Python version (>= 3.11)": python_ok,
    "Required packages": all(package_status.values()),
    "NumPy operation": numpy_ok,
    "Pandas operation": pandas_ok,
    "CSV read/write": csv_ok,
    "Matplotlib execution": plot_ok,
    "scikit-learn operation": sklearn_ok,
}
print("=== ENVIRONMENT CHECK SUMMARY ===")
for title, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL'}  {title}")
if not venv_ok:
    print("WARNING  Notebook kernel does not appear to use a virtual environment.")
print("\nRESULT:", "READY FOR LAB 1" if all(checks.values()) else "SETUP NEEDS ATTENTION")
print("Please also confirm that the Matplotlib chart is visible.")
````

## Troubleshooting

**A package is missing:** Open a terminal in the repository root and reinstall the dependencies:

**Windows (PowerShell or Command Prompt):**

```powershell
.venv\Scripts\python.exe -m pip install -r requirements.txt
```

**Linux:**

```bash
.venv/bin/python -m pip install -r requirements.txt
```

**Wrong Python executable / wrong notebook kernel:** Launch JupyterLab using the virtual environment rather than a global installation:

**Windows:**

```powershell
.venv\Scripts\python.exe -m jupyter lab
```

**Linux:**

```bash
.venv/bin/python -m jupyter lab
```

Then restart the notebook kernel and run all cells again.

**Virtual environment could not be created on Ubuntu/Debian:** Your Python installation may lack `venv`. On a personal laptop with administrator privileges, install the appropriate `python3-venv` package. On university computers, ask the lab administrator for assistance.

**CSV test fails:** Check whether your account can write temporary files and whether disk space is available.

**Plot not visible:** Make sure the notebook is connected to a Python kernel and re-run the plotting cell.

**Keep the error message:** If an issue persists, show the instructor the failing cell and full error message.
